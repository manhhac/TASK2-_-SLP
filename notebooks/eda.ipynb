# Exploratory Data Analysis - Task 2 ASR VI + MT EN

```python
import json
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from collections import Counter

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 6)
```

## Load Manifest Data

```python
def load_manifest(path):
    rows = []
    with open(path, 'r', encoding='utf-8') as f:
        for line in f:
            rows.append(json.loads(line))
    return pd.DataFrame(rows)

# Update paths as needed
train_df = load_manifest('data/manifests/train.jsonl')
dev_df = load_manifest('data/manifests/dev.jsonl')
test_df = load_manifest('data/manifests/test.jsonl')

print(f"Train: {len(train_df)} samples")
print(f"Dev: {len(dev_df)} samples")
print(f"Test: {len(test_df)} samples")
```

## Duration Distribution

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, df, name in zip(axes, [train_df, dev_df, test_df], ['Train', 'Dev', 'Test']):
    ax.hist(df['duration'], bins=50, alpha=0.7, edgecolor='black')
    ax.set_xlabel('Duration (seconds)')
    ax.set_ylabel('Count')
    ax.set_title(f'{name} - Duration Distribution')
    ax.axvline(df['duration'].mean(), color='red', linestyle='--', label=f'Mean: {df["duration"].mean():.2f}s')
    ax.legend()

plt.tight_layout()
plt.show()
```

## Language Distribution

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, df, name in zip(axes, [train_df, dev_df, test_df], ['Train', 'Dev', 'Test']):
    lang_counts = df['language'].value_counts()
    ax.bar(lang_counts.index, lang_counts.values, alpha=0.7)
    ax.set_xlabel('Language')
    ax.set_ylabel('Count')
    ax.set_title(f'{name} - Language Distribution')
    for i, v in enumerate(lang_counts.values):
        ax.text(i, v, str(v), ha='center', va='bottom')

plt.tight_layout()
plt.show()
```

## Text Length Distribution

```python
train_df['text_vi_len'] = train_df['text_vi'].str.len()
train_df['text_en_len'] = train_df['text_en'].str.len()

fig, axes = plt.subplots(1, 2, figsize=(16, 5))

axes[0].hist(train_df['text_vi_len'], bins=50, alpha=0.7, edgecolor='black', color='blue')
axes[0].set_xlabel('Character Count')
axes[0].set_ylabel('Count')
axes[0].set_title('Vietnamese Text Length Distribution')

axes[1].hist(train_df['text_en_len'], bins=50, alpha=0.7, edgecolor='black', color='green')
axes[1].set_xlabel('Character Count')
axes[1].set_ylabel('Count')
axes[1].set_title('English Text Length Distribution')

plt.tight_layout()
plt.show()
```

## Top Words Frequency

```python
def get_top_words(texts, n=20):
    all_words = ' '.join(texts).split()
    word_freq = Counter(all_words)
    return word_freq.most_common(n)

top_vi_words = get_top_words(train_df['text_vi'].fillna(''))
top_en_words = get_top_words(train_df['text_en'].fillna(''))

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

words_vi, counts_vi = zip(*top_vi_words)
axes[0].barh(range(len(words_vi)), counts_vi)
axes[0].set_yticks(range(len(words_vi)))
axes[0].set_yticklabels(words_vi)
axes[0].set_xlabel('Frequency')
axes[0].set_title('Top 20 Vietnamese Words')
axes[0].invert_yaxis()

words_en, counts_en = zip(*top_en_words)
axes[1].barh(range(len(words_en)), counts_en)
axes[1].set_yticks(range(len(words_en)))
axes[1].set_yticklabels(words_en)
axes[1].set_xlabel('Frequency')
axes[1].set_title('Top 20 English Words')
axes[1].invert_yaxis()

plt.tight_layout()
plt.show()
```

## Speaker Distribution

```python
speaker_counts = train_df['speaker_id'].value_counts()
print(f"Total unique speakers: {len(speaker_counts)}")
print(f"\nTop 10 speakers by sample count:")
print(speaker_counts.head(10))

plt.figure(figsize=(12, 5))
plt.hist(speaker_counts.values, bins=50, alpha=0.7, edgecolor='black')
plt.xlabel('Samples per Speaker')
plt.ylabel('Number of Speakers')
plt.title('Speaker Sample Count Distribution')
plt.axvline(speaker_counts.median(), color='red', linestyle='--', label=f'Median: {speaker_counts.median():.0f}')
plt.legend()
plt.show()
```

## Gender Distribution

```python
if 'gender' in train_df.columns:
    gender_counts = train_df['gender'].value_counts()
    print(f"Gender distribution:")
    print(gender_counts)
    
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    for ax, df, name in zip(axes, [train_df, dev_df, test_df], ['Train', 'Dev', 'Test']):
        if 'gender' in df.columns:
            gender_split = df['gender'].value_counts()
            ax.bar(gender_split.index, gender_split.values, alpha=0.7)
            ax.set_xlabel('Gender')
            ax.set_ylabel('Count')
            ax.set_title(f'{name} - Gender Distribution')
            for i, v in enumerate(gender_split.values):
                ax.text(i, v, str(v), ha='center', va='bottom')
    plt.tight_layout()
    plt.show()
else:
    print("Gender information not available in dataset")
```

## Region Distribution

```python
if 'region' in train_df.columns:
    region_counts = train_df['region'].value_counts()
    print(f"Region distribution:")
    print(region_counts)
    
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    for ax, df, name in zip(axes, [train_df, dev_df, test_df], ['Train', 'Dev', 'Test']):
        if 'region' in df.columns:
            region_split = df['region'].value_counts()
            ax.bar(region_split.index, region_split.values, alpha=0.7)
            ax.set_xlabel('Region')
            ax.set_ylabel('Count')
            ax.set_title(f'{name} - Region Distribution')
            ax.tick_params(axis='x', rotation=45)
            for i, v in enumerate(region_split.values):
                ax.text(i, v, str(v), ha='center', va='bottom')
    plt.tight_layout()
    plt.show()
else:
    print("Region information not available in dataset")
```
